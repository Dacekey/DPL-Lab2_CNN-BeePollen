# Bee Pollen CNN — Phase 2: corrected-label experiment

Attach two Kaggle Inputs before running: the original Bee Pollen dataset and a dataset containing a manually reviewed `corrections.csv`. Select **GPU T4 x2** and enable Internet.

```python
# Set CORRECTIONS_CSV to the exact path displayed by Kaggle's Input panel.
DATA_DIR = '/kaggle/input/datasets/ivanfel/honey-bee-pollen'
CORRECTIONS_CSV = '/kaggle/input/datasets/REPLACE_WITH_YOUR_USERNAME/REPLACE_WITH_CORRECTIONS_SLUG/corrections.csv'

REPOSITORY_URL = 'https://github.com/Dacekey/DPL-Lab2_CNN-BeePollen.git'
REPOSITORY_DIR = '/kaggle/working/DPL-Lab2_CNN-BeePollen'
EPOCHS = 25
IMAGE_SIZE = 128
BATCH_PER_REPLICA = 32
SEED = 42
```

```python
!nvidia-smi

import tensorflow as tf
print('TensorFlow GPUs:', tf.config.list_physical_devices('GPU'))
```

```python
import subprocess
from pathlib import Path

assert Path(DATA_DIR).is_dir(), f'DATA_DIR not found: {DATA_DIR}'
assert 'REPLACE_WITH' not in CORRECTIONS_CSV, 'Set CORRECTIONS_CSV first.'
assert Path(CORRECTIONS_CSV).is_file(), f'CORRECTIONS_CSV not found: {CORRECTIONS_CSV}'

if not Path(REPOSITORY_DIR).exists():
    subprocess.run(['git', 'clone', '--depth', '1', REPOSITORY_URL, REPOSITORY_DIR], check=True)
else:
    print('Repository already cloned:', REPOSITORY_DIR)
```

```python
import shlex

command = [
    'python', f'{REPOSITORY_DIR}/src/train_lab.py',
    '--data-dir', DATA_DIR,
    '--corrections-csv', CORRECTIONS_CSV,
    '--output-dir', f'{REPOSITORY_DIR}/outputs',
    '--epochs', str(EPOCHS),
    '--image-size', str(IMAGE_SIZE),
    '--batch-per-replica', str(BATCH_PER_REPLICA),
    '--seed', str(SEED),
    '--run-mode', 'corrected',
]
print('Running:', ' '.join(shlex.quote(x) for x in command))
subprocess.run(command, check=True)
```

```python
from pathlib import Path
import pandas as pd

output_dir = Path(REPOSITORY_DIR) / 'outputs'
package = output_dir / 'packages' / 'improved_corrected.zip'
assert package.is_file(), f'Expected package missing: {package}'
print(f'Download this file after Save Version: {package} ({package.stat().st_size / 1024 / 1024:.1f} MB)')
display(pd.read_csv(output_dir / 'metrics_summary.csv'))
```
